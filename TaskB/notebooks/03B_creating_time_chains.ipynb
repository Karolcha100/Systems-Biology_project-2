```python
def deep_dive(current_depth: int, max_depth: int, image_T: int, df: pd.DataFrame) -> list[list[int|float]]:
    to_return_list: list[list[int|float]] = []

    if current_depth == max_depth:
        return [[]]

    for track_id_t0 in df.index.get_level_values(1).unique():
        list_of_erk_diffs: list[list[int|float]] = [
            [track_id_t0, erk_neigh_diff] + deep_dive_on_sel_node(
                track_id_t1, current_depth, max_depth, image_T, 
                df.loc[pd.IndexSlice[image_T+1:, :, :], :]
            )
            for (_, _, track_id_t1), erk_neigh_diff in df.loc[pd.IndexSlice[image_T+current_depth, track_id_t0, :], "ERK_neigh_diff"].to_dict().items()
        ]


        to_return_list.extend(
                list_of_erk_diffs
            )
        
    return to_return_list
    
```

```python
def find_chains(df: pd.DataFrame, chains_depth: int) -> pd.DataFrame:
    given_columns: list[str] = []
    for depth in range(chains_depth):
        given_columns.append(f"track_id_t{depth}")
        given_columns.append(f"ERK_diff_t{depth}")


    df_chained = pd.DataFrame(
        columns=given_columns 
    )

    for image_T in range(0, len(df.index.get_level_values(0).unique().to_list()) - chains_depth):

        deep_dive_raw: list[list[int|float]] = deep_dive(0, 5, image_T, df.loc[pd.IndexSlice[image_T:image_T+chains_depth, :, :], :])

        deep_dive_df: pd.DataFrame = pd.DataFrame(
                deep_dive_raw, 
                columns=given_columns,
        )

        df_chained = pd.concat([
            df_chained, 
            deep_dive_df
        ])

    return df_chained
```

```python
import pandas as pd
import numpy as np

def find_chains(df: pd.DataFrame, chains_depth: int = 5) -> pd.DataFrame:
    """
    Znajduje łańcuchy propagacji ERK długości `chains_depth`.
    
    Oczekiwany MultiIndex df: (image_T, track_id, neighbour_id)
    Oczekiwana kolumna: ERK_neigh_diff
    """
    # --- Krok 1: spłaszcz indeks do kolumn, zostaw tylko co potrzebne ---
    base = (
        df[["ERK_neigh_diff"]]
        .reset_index()
        .rename(columns={
            df.index.names[0]: "image_T",
            df.index.names[1]: "track_id",
            df.index.names[2]: "neighbour_id",
        })
    )

    # --- Krok 2: zbuduj tabelę krawędzi (track -> neighbour) per timepoint ---
    # Każdy wiersz: "w timepoint T, komórka track_id ma sąsiada neighbour_id z ERK_diff"
    edges = base[["image_T", "track_id", "neighbour_id", "ERK_neigh_diff"]].copy()

    # --- Krok 3: iteracyjny self-join po kolejnych timepoint'ach ---
    # Startujemy od krawędzi w t=0 (względne)
    chains = edges.rename(columns={
        "track_id":      "track_id_t0",
        "neighbour_id":  "track_id_t1",
        "ERK_neigh_diff": "ERK_diff_t0",
    })

    for depth in range(1, chains_depth):
        # Krawędzie przesunięte o +1 timepoint
        next_edges = edges.copy()
        next_edges["image_T"] = next_edges["image_T"] - depth  # wyrównaj do bazowego T

        next_edges = next_edges.rename(columns={
            "track_id":       f"track_id_t{depth}",
            "neighbour_id":   f"track_id_t{depth+1}",
            "ERK_neigh_diff": f"ERK_diff_t{depth}",
        })

        # Join: koniec poprzedniego łańcucha == początek następnej krawędzi
        chains = chains.merge(
            next_edges,
            left_on=["image_T", f"track_id_t{depth}"],
            right_on=["image_T", f"track_id_t{depth}"],
            how="inner",
        )

    # --- Krok 4: usuń pomocnicze kolumny i posortuj ---
    col_order = ["image_T"]
    for d in range(chains_depth):
        col_order.append(f"track_id_t{d}")
        col_order.append(f"ERK_diff_t{d}")
    col_order.append(f"track_id_t{chains_depth}")  # ostatni węzeł łańcucha

    chains = chains[col_order].sort_values("image_T").reset_index(drop=True)

    return chains
```
