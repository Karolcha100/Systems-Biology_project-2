# Viewing Data

```python
import matplotlib.pyplot as plt
```

```python
def plot_data(df: pd.DataFrame, labels: list[int]|None = None, alpha = 0.05) -> None:
    fig, ax = plt.subplots(1, figsize=(10, 10))

    if labels is None:
        ax.scatter(
            df[df.columns[0]].to_numpy(),
            df[df.columns[1]].to_numpy(),
            marker='.',
            alpha = alpha,
        )
    else:
        for lab_idx in labels:
            ax.scatter(
                df[df.columns[0]].to_numpy(),
                df[df.columns[1]].to_numpy(),
                marker='.',
                alpha = alpha,
                label = f"lab{lab_idx}",
            )

    ax.set_title("Data Visualized")

    ax.set_xlabel(df.columns[0])
    ax.set_xlabel(df.columns[1])

    ax.legend()
```

```python
plot_data(
    df_graphs_flatten["ERK_me_diff ERK_neigh_diff".split(" ")], 
    labels=[lab_idx for lab_idx in df_graphs_flatten["label"].value_counts().to_dict().keys() if lab_idx not in {-1}],
    alpha = 0.01
)
```
